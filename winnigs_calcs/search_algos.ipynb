```python
import numpy as np

def random_solution(n):
    return np.random.randint(0,2,size=n)

def mutate(sol, rate=0.05):
    out = sol.copy()
    mask = np.random.rand(len(sol)) < rate
    out[mask] = 1 - out[mask]
    return out

def crossover(a,b):
    cut = np.random.randint(1,len(a))
    return np.concatenate([a[:cut], b[cut:]])

def evaluate(sol, probs, odds, fstar, sims=5000):
    idx = sol==1
    if idx.sum()==0:
        return -999
    payoff = []
    for _ in range(sims):
        win = np.random.rand(idx.sum()) < probs[idx]
        profit = np.sum(win*(fstar[idx]*odds[idx]) - fstar[idx])
        payoff.append(profit)
    return np.mean(payoff)

def GA(probs, odds, fstar, pop=30, gens=50):
    n = len(probs)
    population = [random_solution(n) for _ in range(pop)]

    for g in range(gens):
        fitness = np.array([evaluate(s, probs, odds, fstar) for s in population])
        elite = population[np.argmax(fitness)]

        new_pop = [elite.copy()]
        for _ in range(pop-1):
            a,b = np.random.choice(population, 2)
            child = mutate(crossover(a,b))
            new_pop.append(child)
        population = new_pop
    
    best = max(population, key=lambda s: evaluate(s, probs, odds, fstar))
    return best
```

```python
def find_duplicates(df):
    duplicates = (
        df[df.duplicated(
            subset=['fighter_red', 'fighter_blue', 'date'],
            keep=False
        )]
        .sort_values(['date', 'fighter_red', 'fighter_blue'])
        .reset_index(drop=True)
    )
    print(duplicates[['fighter_red', 'fighter_blue']])

find_duplicates(df_rolling)
```

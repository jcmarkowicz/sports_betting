```python
import numpy as np 
import pandas as pd 
from collections import defaultdict
 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df_v2 = pd.read_csv(r'C:\Users\jcmar\my_files\SportsBetting\data\entire_odds_stats_2025-11-23.csv')
df_v2 = df_v2.sort_values('date').reset_index(drop=True)
df_v2 = df_v2[df_v2['winner']!=2].reset_index(drop=True)
# df_v2 = df_v2.dropna().reset_index(drop=True)

# df_non_rolling = pd.read_csv(r'C:\Users\jcmar\my_files\SportsBetting\data\new_combined.csv')
# print(df_v2.shape, df_non_rolling.shape)
```

```python
fighters_red = df_v2[['fighter_red', 'elo_red','date']].rename(columns={'fighter_red': 'fighter', 'elo_red': 'elo'})
fighters_blue = df_v2[['fighter_blue', 'elo_blue', 'date']].rename(columns={'fighter_blue': 'fighter', 'elo_blue': 'elo'})
print(fighters_blue.columns)

all_fighters = pd.concat([fighters_red, fighters_blue], ignore_index=True)
all_fighters = all_fighters.sort_values('date')   # or whatever your chronological column is

top_fighters = all_fighters.groupby('fighter')['elo'].last().sort_values(ascending=False)


top_10_fighters = top_fighters.head(50)
top_10_fighters
```

```python
df_test = pd.read_csv(r'C:\Users\jcmar\my_files\SportsBetting\data\new_combined.csv')
```

```python
red_ratings = df_test[['fighter_red', 'glicko_red']].rename(
    columns={'fighter_red': 'fighter', 'glicko_red': 'rating'}
)
blue_ratings = df_test[['fighter_blue', 'glicko_blue']].rename(
    columns={'fighter_blue': 'fighter', 'glicko_blue': 'rating'}
)

all_ratings = pd.concat([red_ratings, blue_ratings], ignore_index=True)

# Find peak rating per fighter
peak_ratings = all_ratings.groupby('fighter')['rating'].max().reset_index()

# Sort descending
peak_ratings = peak_ratings.sort_values('rating', ascending=False).reset_index(drop=True)

# Show top 10 fighters by peak rating
print(peak_ratings.head(10))
```

```python
df_test['winner'] = df_test['winner']
df_test['choice_elo'] = np.where(df_test['elo_red'] > df_test['elo_blue'], 1, 0)

df_test['choice_correct'] = np.where(df_test['choice_elo']==df_test['winner'], 1, 0)
print(df_test['choice_correct'].mean())
```
